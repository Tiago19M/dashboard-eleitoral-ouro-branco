##Mapas individuais - Primeiro Turno

Os seis mapas seguem a mesma estrutura e lógica de design visando facilitar a comparação entre eles. Abaixo, são descritas as escolhas de representação bem como a justificativa para tais e, logo em seguida, seguem as análises dos padrões observados em cada mapa individualmente.



###Justificativa de estrutura/design

Unidade espacial: O TSE não disponibiliza polígonos das seções eleitorais, apenas as coordenadas dos locais de votação (22 no total). Assim, para determinar as áreas correspondentes à cada um dos 22 pontos representativos dos locais de votação, adotou-se um diagrama de Voronoi recortado pelo limite municipal (fornecido pelo IBGE). Nesse modelo, cada ponto do território de Ouro Branco foi atribuído ao local de votação mais próximo e, portanto, as representações se tratam de uma aproximação geométrica, e não da área real de captação de eleitores.


---



Variável representada: Cada mapa mostra o percentual de votos válidos do primeiro turno (excluindo brancos e nulos) do candidato em questão em cada local. Optou-se pelo uso do percentual de votos para que diferentes tamanhos de locais de votação não distorçam a preferência de candidato de cada região definida.


---



Candidatos: Dentre os 12 candidatos com voto na cidade, escolheu-se os 6 candidatos mais votados no município (aqueles que possuíam porcentagem de votos válidos superior a 0,1%)


---



Paleta e classes: Foram usadas paletas sequenciais com cores-base que remetem ao partido do candidato, onde, tons mais claros indicam menor percentual enquanto tons mais fortes um maior percentual de votos. A associação das regiões à cor do partido é estratégica e facilita a associação visual do observador. Sobre as classes, considerou-se 5 para os 22 polígonos, levando em conta que mais classes poderia deixar poucos locais por faixa e tornaria os tons mais difíceis de distinguir, enquanto menos poderiam esconder variações relevantes.


---



Escala: Cada mapa tem sua própria escala, que leva em conta a quantidade de votos obtidos pelo candidato em questão. Com isso, garante-se que não haja perca de detalhes nas variações de votos obtidos por um candidato em específico, apesar de impossibilitar a comparação direta entre diferentes candidatos.


---



Componentes do folium: folium.Map (mapa base), folium.Choropleth (preenchimento por classes, com legenda) e GeoJsonTooltip (caixa ao passar o mouse com local de votação, bairro e percentual exato, estilizada com CSS). A caixa traz os percentuais exatos, que a cor sozinha não permite ler.

```python
df_eleitorado = pd.read_csv(CAMINHO_BASE / "eleitorado_ouro_branco_2026.csv")
df_eleitorado.head(5)
```

```python
chave = ["NR_TURNO", "NR_ZONA", "NR_SECAO"]

colunas_loc = chave + ["NM_BAIRRO", "NR_LATITUDE", "NR_LONGITUDE", "QT_ELEITOR_SECAO"]

#Para evitar colunas duplicadas ao executar a célula duas vezes:
df_pandas = df_pandas.drop(
    columns=["NM_BAIRRO", "NR_LATITUDE", "NR_LONGITUDE", "QT_ELEITOR_SECAO"],
    errors="ignore"
)

#Unindo as colunas de localização das zonas eleitorais ao df_pandas
df_pandas = df_pandas.merge(
    df_eleitorado[colunas_loc].drop_duplicates(subset=chave),
    on=chave,
    how="left"
)
```

#Etapa 3

```python
import geopandas as gpd
from shapely.geometry import MultiPoint
from shapely.ops import voronoi_diagram

# Definição dos polígonos representativos dos locais de votação (Voronoi)

locais = df_eleitorado.drop_duplicates(subset="NR_LOCAL_VOTACAO")
locais = locais[["NR_LOCAL_VOTACAO", "NM_LOCAL_VOTACAO", "NM_BAIRRO", "NR_LATITUDE", "NR_LONGITUDE"]].copy()

pontos = gpd.GeoDataFrame(
    locais, #dataFrame contendo os locais de votação
    geometry=gpd.points_from_xy(locais["NR_LONGITUDE"], locais["NR_LATITUDE"]), #Cria a coluna 'geometry' do gdf com as coordenadas dos locais de votação
    crs=4326 #Indica que a latitude e longitude está em graus
)

limite = gpd.read_file(CAMINHO_BASE / "ouro_branco_limite.geojson")

SIST_REF = 31983 #sistema de referência de coordenadas que permite trabalhar com distâncias em metros
pontos_metros = pontos.to_crs(SIST_REF)  #to_crs converte as coordenadas pro novo sistema
limite_metros = limite.to_crs(SIST_REF)

lim = limite_metros.geometry.union_all() #une as geometrias numa só
multi = MultiPoint(list(pontos_metros.geometry)) #reúne os 22 pontos num objeto aceitável pela função do Voronoi
voronoi = voronoi_diagram(multi, envelope=lim.buffer(5000))

celulas = gpd.GeoDataFrame(geometry=list(voronoi.geoms), crs=SIST_REF) #cria uma tabela só com a coluna de geometrias do mapa

voronoi_loc = gpd.sjoin(celulas, pontos_metros, predicate="contains") #utiliza junção espacial (sjoin) para correlacionar geometrias aos locais de votação correspondentes (pontos)
voronoi_loc["geometry"] = voronoi_loc.geometry.intersection(lim) #limita os polígonos à área do município de Ouro Branco, atualizando-os.

voronoi_loc = voronoi_loc.to_crs(4326).reset_index(drop=True) #faz a conversão para graus novamente (o Folium trabalha em graus)
voronoi_loc["CHAVE"] = voronoi_loc["NR_LOCAL_VOTACAO"].astype(str) #cria uma chave de união de tipo comum que ligará os polígonos aos percentuais de votação por local
```

```python
#Cálculo de votos por local e candidato
prim_turno = df_pandas[df_pandas["NR_TURNO"] == 1].copy()
nomes = prim_turno["NM_VOTAVEL"]
prim_turno = prim_turno[~(nomes.str.contains("BRANCO") | nomes.str.contains("NULO"))]

votos_por_local = prim_turno.groupby(["NR_LOCAL_VOTACAO", "NM_VOTAVEL"])["QT_VOTOS"].sum().reset_index()

#Cálculo do percentual de votos por local e candidato
total_local = votos_por_local.groupby("NR_LOCAL_VOTACAO")["QT_VOTOS"].transform("sum")
votos_por_local["PERCENTUAL"] = votos_por_local["QT_VOTOS"] / total_local*100


votos_por_local["CHAVE"] = votos_por_local["NR_LOCAL_VOTACAO"].astype(str)
```

```python
#Estilo css usado nos tooltips dos mapas gerados
css = """
  <style>
  .tooltip-eleitoral{
    background: #B8BDB5 !important;
    border: 1px solid #0F0E0E !important;
    padding: 12px 14px !important;
    color: #000000 !important;
    font-family: Arial, sans-serif !important;
    font-size: 12px !important;
    line-height: 1.5 !important
  }


.tooltip-eleitoral td {
    padding: 4px 2px;
    border-bottom: 1px solid #334155;
}

/* Última linha sem separador */
.tooltip-eleitoral tr:last-child td {
    border-bottom: none;
}

/* Destaque dos nomes dos candidatos */
.tooltip-eleitoral tr:nth-child(4) td {
    color: #60a5fa;
    font-weight: bold;
}

.tooltip-eleitoral tr:nth-child(5) td {
    color: #f87171;
    font-weight: bold;
}
  </style>
"""
```

```python
candidatos = [
    "FLAVIO NANTES BOLSONARO",
    "LUIZ INÁCIO LULA DA SILVA",
    "AUGUSTO JORGE CURY",
    "RENAN ANTONIO FERREIRA DOS SANTOS",
    "RONALDO RAMOS CAIADO",
    "ROMEU ZEMA NETO",
]

paletas = {
    "FLAVIO NANTES BOLSONARO" : "Blues",
    "LUIZ INÁCIO LULA DA SILVA" : "Reds",
    "AUGUSTO JORGE CURY" : "Greys",
    "RENAN ANTONIO FERREIRA DOS SANTOS" : "YlOrBr",
    "RONALDO RAMOS CAIADO" : "Greens",
    "ROMEU ZEMA NETO" : "YlOrBr",
}

mapas = {}

for candidato in candidatos:
  dados = votos_por_local[votos_por_local["NM_VOTAVEL"] == candidato][["CHAVE", "PERCENTUAL"]]
  dados = voronoi_loc[["CHAVE"]].merge(dados, on="CHAVE", how="left").fillna({"PERCENTUAL" : 0})

  mapa = folium.Map(
      location=[-20.52, -43.69],
      zoom_start=12,
      tiles="https://tile.openstreetmap.de/{z}/{x}/{y}.png",
      attr="&copy; OpenStreetMap contributors",
  )

  folium.Choropleth(
      geo_data=voronoi_loc[["CHAVE", "geometry"]],
      name=candidato,
      data=dados,
      columns=["CHAVE", "PERCENTUAL"],
      key_on="feature.properties.CHAVE",
      fill_color=paletas[candidato],
      fill_opacity=0.7,
      line_opacity=0.3,
      bins=5,
      legend_name=f"% dos votos válidos - {candidato}",
  ).add_to(mapa)

  #Lógica do tooltip nos gráficos
  tabela_atual = votos_por_local[votos_por_local["NM_VOTAVEL"] == candidato][["NR_LOCAL_VOTACAO", "PERCENTUAL"]].copy()
  tabela_atual["PERCENTUAL"] = tabela_atual["PERCENTUAL"].round(2)
  tabela_atual["CHAVE"] = tabela_atual["NR_LOCAL_VOTACAO"].astype(str)

  geo_t1 = voronoi_loc[["CHAVE", "NM_LOCAL_VOTACAO", "NM_BAIRRO", "geometry"]].merge(tabela_atual, on="CHAVE", how="left")

  tooltip_t1 = folium.GeoJsonTooltip(
      fields = [
          "NM_LOCAL_VOTACAO",
          "NM_BAIRRO",
          "PERCENTUAL"
      ],
      aliases = [
          "Local de votação:",
          "Bairro:",
          "Votos (%):"
      ],
      sticky = True,
      class_name = "tooltip-eleitoral"
  )

  folium.GeoJson(
    geo_t1,
    tooltip = tooltip_t1,
    style_function=lambda feature: {
        "fillOpacity": 0,
        "color": "transparent",
        "weight": 0
    }).add_to(mapa)

  mapa.get_root().html.add_child(folium.Element(css))
  mapas[candidato] = mapa
```

```python
segundo_turno = df_pandas[df_pandas["NR_TURNO"] == 2].copy()
nomes = segundo_turno["NM_VOTAVEL"]
segundo_turno = segundo_turno[~(nomes.str.contains("BRANCO")) & ~(nomes.str.contains("NULO"))] #Elimina-se os votos brancos e nulos

votos_por_local_2 = segundo_turno.groupby(["NR_LOCAL_VOTACAO", "NM_VOTAVEL"])["QT_VOTOS"].sum().reset_index()

total_local_2 = votos_por_local_2.groupby("NR_LOCAL_VOTACAO")["QT_VOTOS"].transform("sum")
votos_por_local_2["PERCENTUAL"] = (votos_por_local_2["QT_VOTOS"] / total_local_2*100)

#Para os dados do gráfico, considerou-se a porcentagem de votos por local de Flávio Bolsonaro para aproveitar a paleta 'RdBu', que associa azul à valores mais altos
#e vermelho aos valores mais baixos. Isso permite associar as regiões às cores do partido dominante (com mais votos no local)
dados_votos = votos_por_local_2[votos_por_local_2["NM_VOTAVEL"] == "FLAVIO NANTES BOLSONARO"][["NR_LOCAL_VOTACAO", "PERCENTUAL"]].copy()
dados_votos["CHAVE"] = dados_votos["NR_LOCAL_VOTACAO"].astype(str)

#É feito um merge associando dados de votação (percentuais de voto) às chaves de cada polígono de votação definidos
dados_mapa = voronoi_loc[["CHAVE"]].merge(dados_votos, on = "CHAVE", how = "left")
candidato_a = "FLAVIO NANTES BOLSONARO"
candidato_b = "LUIZ INÁCIO LULA DA SILVA"

tabela_a = votos_por_local_2[votos_por_local_2["NM_VOTAVEL"] == candidato_a][["NR_LOCAL_VOTACAO", "PERCENTUAL"]]
tabela_a = tabela_a.rename(columns={"PERCENTUAL" : "PERCENTUAL_A"})

tabela_b = votos_por_local_2[votos_por_local_2["NM_VOTAVEL"] == candidato_b][["NR_LOCAL_VOTACAO", "PERCENTUAL"]]
tabela_b = tabela_b.rename(columns={"PERCENTUAL" : "PERCENTUAL_B"})

tabela_t2 = tabela_a.merge(tabela_b, on="NR_LOCAL_VOTACAO", how="outer") #O parâmetro 'outer' preserva todas as linhas, mesmo que um candidato não tenha recebido votos em algum local.
tabela_t2 = tabela_t2.fillna(0) #Se recebeu 0 votos, recebe 0 na porcentagem
tabela_t2["CHAVE"] = tabela_t2["NR_LOCAL_VOTACAO"].astype(str)

cand_a_lidera = tabela_t2["PERCENTUAL_A"] > tabela_t2["PERCENTUAL_B"]
tabela_t2["LIDER"] = np.where(cand_a_lidera, candidato_a, candidato_b)
tabela_t2["PERCENTUAL_LIDER"] = np.where(cand_a_lidera, tabela_t2["PERCENTUAL_A"], tabela_t2["PERCENTUAL_B"])

#Arredondamentos
tabela_t2["PERCENTUAL_LIDER"] = tabela_t2["PERCENTUAL_LIDER"].round(2)
tabela_t2["PERCENTUAL_A"] = tabela_t2["PERCENTUAL_A"].round(2)
tabela_t2["PERCENTUAL_B"] = tabela_t2["PERCENTUAL_B"].round(2)

geo_t2 = voronoi_loc[["CHAVE", "NM_LOCAL_VOTACAO", "NM_BAIRRO", "geometry"]].merge(tabela_t2, on="CHAVE", how="left")
```
